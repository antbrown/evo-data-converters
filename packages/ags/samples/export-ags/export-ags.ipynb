# Export Evo Downhole Collection to AGS file

Use this notebook to login to Evo and export a Downhole Collection to AGS 4.1 file.

At the moment supported tables for export include:

- `LOCA` - Collar details for each downhole
- `SCPG` - General SCPT data from hole collars

Survey data

- `SCPT` - CPT distance table
- `GEOL` - Geology interval table
- `SCPP` - Derived information from the SCPT table

Static tables

- `ABBR` - Abbreviations used within an AGS file
- `UNIT` - Unit abbreviations and their descriptors
- `TYPE` - Type abbreviations and their descriptors

Metadata

- `PROJ` - Information about the overall project
- `TRAN` - File transfer information

```python
import os

from evo.notebooks import ServiceManagerWidget

client_id = os.getenv("EVO_CLIENT_ID", "")
base_uri = os.getenv("EVO_BASE_URI", "")
discovery_url = os.getenv("EVO_DISCOVERY_URL", "")

manager = await ServiceManagerWidget.with_auth_code(
    client_id=client_id, base_uri=base_uri, discovery_url=discovery_url
).login()
```

Find the UUID and version number from Evo for the target object.

Choose a filepath to export the object to (relative to this directory).

```python
from uuid import UUID

from evo.data_converters.ags.exporter import export_ags
from evo.data_converters.common.evo_client import EvoObjectMetadata

object_meta = EvoObjectMetadata(object_id=UUID(""), version_id="")

export_path = "ags_export.ags"

await export_ags(filepath=export_path, objects=[object_meta], service_manager_widget=manager)
```

You can use the cell below to validate the AGS output.

```python
from pprint import pp

from python_ags4.AGS4 import check_file

validation_results = check_file(export_path)

pp(validation_results)
```
